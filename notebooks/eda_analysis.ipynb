```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load cleaned data
df = pd.read_csv('../data/cleaned_data.csv')

# Statistical health check 
print(df.describe())
print(df.info())
```

```python
# Correlation Heatmap 
plt.figure(figsize=(10, 8))
corr_matrix = df.corr(numeric_only=True) 
sns.heatmap(corr_matrix, annot=True, cmap='magma', fmt='.2f')
plt.title('Feature Correlation Heatmap (Halloween Theme)')
plt.show()

# Distribution Plots
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df['CGPA'], kde=True, color='#FF7518') 
plt.title('CGPA Distribution (Cleaned)')

plt.subplot(1, 2, 2)
sns.histplot(df['IQ'], kde=True, color='#710193') 
plt.title('IQ Score Distribution (Cleaned)')
plt.show()

# Box Plots: Placement vs Academic Performance
plt.figure(figsize=(10, 6))
sns.boxplot(x='Placement', y='CGPA', data=df, palette=['#710193', '#FF7518'])
plt.title('Academic Performance vs Placement Status')
plt.show()
```

```python
from scipy.stats import skew

numerical_cols = ['CGPA', 'IQ', 'Academic_Efficiency']
for col in numerical_cols:
    print(f"{col} Skewness: {skew(df[col].dropna()):.2f}")
```

```python
from matplotlib.colors import LinearSegmentedColormap
colors = ["#000000", "#710193", "#FF7518"]
halloween_cmap = LinearSegmentedColormap.from_list("halloween", colors)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Using a Violin plot to show the probability density of CGPA for Placed vs Non-Placed
plt.figure(figsize=(10, 6))
sns.violinplot(x='Placement', y='CGPA', data=df, inner="quart", palette="magma")
plt.title('CGPA Density Distribution by Placement Status') 
plt.show()

# Standard Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap=halloween_cmap, fmt='.2f') 
plt.title('Correlation Matrix (Including New Engineered Features)')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Comparison: Academic Performance vs Placement
plt.figure(figsize=(12, 6))

# Box plot for CGPA 
plt.subplot(1, 2, 1)
sns.boxplot(x='Placement', y='CGPA', data=df, palette='flare')
plt.title('Outlier-Filtered CGPA vs Placement')

# Violin plot for IQ Score  
plt.subplot(1, 2, 2)
sns.violinplot(x='Placement', y='IQ', data=df, palette='crest')
plt.title('IQ Score Distribution by Placement')

plt.tight_layout()
plt.show()
```
